# Hyperparameter Tuning

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("iris")
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop('species' , axis = 1)
y = df['species']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=9)
```

```python
model_knn.fit(x_train, y_train)
```

```python
model_knn.score(x_test,y_test)
```

# SVM

```python
from sklearn.svm import SVC

model_SVM = SVC(C=30 , kernel='rbf' , gamma='auto')
```

```python
model_SVM.fit(x_train , y_train)
```

```python
model_SVM.score(x_test,y_test)
```

# Grid Search CV

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_SVM ), {
    'C' : [1,10,20,30],
    'kernel' : ['rbf', 'linear'],
} , cv=5 , return_train_score=False)
```

```python
classifier.fit(x,y)
```

```python
classifier.cv_results_
```

```python
result = pd.DataFrame(classifier.cv_results_)
result
```

```python
result[['param_C', 'param_kernel', 'mean_test_score']]
```

# Random Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_random = RandomizedSearchCV((model_SVM ), {
    'C' : [1,10,20,30],
    'kernel' : ['rbf', 'linear'],
} ,n_iter = 4, cv = 5 , return_train_score=False)
```

```python
classifier_random.fit(x,y)
```

```python
result = pd.DataFrame(classifier_random.cv_results_)
result
```

```python
result[['param_C', 'param_kernel', 'mean_test_score']]
```
